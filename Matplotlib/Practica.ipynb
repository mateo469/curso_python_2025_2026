```python
#Importar la librerias pandas, matplotlib y numpy
import matplotlib.pyplot as plt
import pandas as pd
import os
# Ver ruta actual
os.chdir(r"C:\Users\monti.UNITY12\OneDrive\Documentos\Curso Python 2026")
```

```python

#Importar un archivo csv llamado accidentes.csv
df = pd.read_csv("accidentes.csv", index_col='NumeroCaso')
df.dropna(inplace=True)
```

```python
#Mostrar los primeros elementos
df.head()
```

```python
#Eliminacion de columna
Eliminada = df.drop(columns=['Estado', 'Personas _no_vehiculos',
                             'Vehiculos automotores','NombreCondado'], inplace=False)
Eliminada
```

```python
#Crear grupos
Eliminada.groupby('Clima').get_group('lluvia')
```

```python
grupo_Clima = Eliminada.groupby('Clima')['Condado'].sum().reset_index()
grupo_Clima = grupo_Clima.sort_values(by='Condado', ascending=False)
grupo_Clima
```

```python
import matplotlib.pyplot as plt

# Ordenar de mayor a menor por 'Condado'
grupo_Clima = grupo_Clima.sort_values(by='Condado', ascending=False)

# Gráfica de barras
plt.figure(figsize=(10,6))
plt.bar(grupo_Clima['Clima'], grupo_Clima['Condado'], color='skyblue')

# Etiquetas y título
plt.xlabel('Clima')
plt.ylabel('Suma de Condados')
plt.title('Suma de Condados por Clima')
plt.legend(grupo_Clima['Clima'], loc='upper right')#Mostrar la leyenda
plt.xticks(rotation=75, ha='right')  # Rotar etiquetas para mejor lectura
plt.tight_layout()

plt.show()
```

```python

#Crear una tabla dinamica para mostrar el clima por conteo de una barra
Tabla1 = pd.pivot_table(Eliminada, values='Condado', index='Clima', aggfunc='max')
Tabla1 = Tabla1.sort_values(by='Condado', ascending=True)#Ordenar
Tabla1
```

```python
#Abrir el archivo de futbolista
deporte = pd.read_csv('futbolista.csv')
deporte
```

```python
#Agrupar ligas por horas entrenada
grupo_liga = deporte.groupby('Liga')['Horas_Entrenando'].sum().reset_index()
#Ordenar
grupo_liga = grupo_liga.sort_values(by='Horas_Entrenando', ascending=False)
plt.Figure(figsize=(10, 5))
#Graficar el grupo liga
plt.bar(grupo_liga['Liga'], grupo_liga['Horas_Entrenando'], color='red')
plt.xlabel('Liga')
plt.show()
```

```python
#Tabla dinamica para mostrar las horas por ligas
tabla2 = pd.pivot_table(deporte, values='Horas_Entrenando', index='Liga', aggfunc='sum')
tabla2
```

```python
#MOstrar las horas en una grafic ade pastel
tabla2.plot(kind='pie', y='Horas_Entrenando', autopct='%1.0f%%', title='Horas por liga', startangle=360)
plt.axis('equal')
plt.show()
```
