##### 🧱 Componentes principales 

###### Elemento	Función plot()	Dibuja una línea entre puntos

**Matplotlib** 

Es una biblioteca de Python para crear gráficos en 2D y 3D. Fue desarrollada en 2002 por John Hunter con el objetivo de replicar las capacidades gráficas de MATLAB en Python.

**🎯 ¿Para qué sirve?**
Sirve para:

Visualizar datos de forma clara y atractiva.

Crear gráficos estáticos, animados e interactivos.

Personalizar cada aspecto del gráfico: colores, estilos, etiquetas, leyendas, etc.

Integrarse fácilmente con NumPy, pandas y otras bibliotecas de análisis de datos.

**⚙️ ¿Cómo funciona?**

Matplotlib se basa en dos conceptos clave:

Figure: el lienzo general donde se dibujan los gráficos.

Axes: los sistemas de coordenadas dentro de la figura.

El módulo más usado es pyplot, que simplifica la creación de gráficos con funciones como plot(), bar(), scatter(), etc.

```python
#Importar la libreria de matplotlib
import matplotlib.pyplot as plt
import numpy as np
#Verificar que version es
#Crear mi primer figura simple
ejex= [1, 2, 3, 4]
ejey= [20, 50, 30, 21]
plt.plot(ejex, ejey)
plt.title('grafico simple')
plt.show()
```

```python
#Ejemplo 2
ejex= ['a', 'b', 'c', 'd', 'e']
ejey= [20, 50, 30, 21, 10]
ejez = [30, 20, 55, 60, 40]
plt.plot(ejex, ejey)
plt.plot(ejex, ejez) 
plt.title('grafico simple 2')
plt.show()
```

```python
#Ejemplo 3
grafica3 = np.array([3, 8, 1, 10])

plt.plot(grafica3, 'o:r')
plt.show()
```
