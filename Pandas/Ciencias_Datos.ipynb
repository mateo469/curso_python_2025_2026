**Tablas Dinamicas**

Crear tablas dinámicas (o pivot tables) con pandas es muy sencillo usando la función pivot_table(). Es muy parecida a las tablas dinámicas de Excel y se usan para resumir y analizar datos agrupados.

## Estructura de una tabla dinamica.

##### pd.pivot_table(data=df, values='columna_valor', index='columna_fila', columns='columna_columna', aggfunc='función')


```python
#Mi primer tabla dinamica basica
#Tabla dinámica del salario promedio por estado
tabla1 = pd.pivot_table(tabla_datos, values='salario', index='estado', aggfunc='mean')
tabla1
```

```python
#Contar cuantas personas hay por estados
tabla2 = pd.pivot_table(tabla_datos, values='nombre', index='estado', aggfunc='count')
tabla2
```

```python
tabla4 = pd.pivot_table(tabla_datos, index='estado', values=['salario', 'edad', 'calificacion'], aggfunc=['mean', 'max', 'min'])
tabla4
#Guardarlo en un archivo csv
#tabla4.to_csv('tabladinamica.csv')
```

```python
import matplotlib.pyplot as plt

tabla = pd.pivot_table(tabla_datos, values='salario', index='estado', aggfunc='mean')

# Crear gráfico de barras
tabla.plot(kind='bar', legend=False)
plt.title('Salario promedio por estado')
plt.ylabel('Salario')
plt.xlabel('Estado')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
#Grafica de lineas.
tabla = pd.pivot_table(tabla_datos, values='calificacion', index='estado', aggfunc='mean')

# Gráfico de líneas
tabla.plot(kind='line', marker='o')
plt.title('Calificación promedio por estado')
plt.ylabel('Calificación')
plt.grid(True)
plt.tight_layout()
plt.show()
```
