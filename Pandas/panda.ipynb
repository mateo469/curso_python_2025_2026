**🔧 Características principales**

DataFrame: estructura bidimensional (como una hoja de Excel) con etiquetas en filas y columnas.

Series: estructura unidimensional, como una columna de datos.

Lectura de datos: desde CSV, Excel, JSON, SQL, etc.

Filtrado y selección: accede a datos específicos con condiciones.

Agrupación y agregación: ideal para análisis estadístico.

Visualización: se integra con Matplotlib para crear gráficos.

```python
#convertir cadenas de textos a fechas
fechas = ["2023-10-1", "2024-11-12"]
```

```python
#convertir a fechas
cumpleaños = pd.to_datetime(fechas)
cumpleaños
```

```python
#Mostrar solo los meses
cumpleaños.month
```

```python
#Mostrar solos los años
cumpleaños.year
```

```python
#Mostraar solos los dias
cumpleaños.dayofweek
```

```python
#Formatear cada fecha en la serie cumpleaños al formato "mm/dd/yy"
cumpleaños.strftime("%m/%d/%y")
```

**Crear un dataframe de una lista de diccionario, estableciendo las columnas.**

```python
#Crear mi lista de diccionario
futbolista = [{'id':1, 'nombre':'Mariana', 'horas_entrenadas': 20, 'posicion': 'delantero', 'liga':1},
              {'id':2, 'nombre':'SoL', 'horas_entrenadas': 10, 'posicion': 'delantero', 'liga':2},
              {'id':3, 'nombre':'Mar', 'horas_entrenadas': 1, 'posicion': 'defensa', 'liga':1},
              {'id':4, 'nombre':'Ofelia', 'horas_entrenadas': 12, 'posicion': 'delantero', 'liga':3},
              {'id':5, 'nombre':'Camila', 'horas_entrenadas': 11, 'posicion': 'portero', 'liga':1}]
```

```python
#Crear el dataframe para verlo en tabla.
futbolista = pd.DataFrame(futbolista).set_index('id')
futbolista
```

```python
#seleccionar columnas nombre y posicion del dataframe.
#acceder a la informacion de 1 columna.
futbolista[['nombre']]
```

```python
#mostrar mas columnas
futbolista[['nombre', 'posicion']]
```

```python
#seleccionar filas del dataframe.
futbolista.loc[[2, 3]]
```

```python
#seleccionar filas del dataframe y mostrar solo las columnas seleccionadas.
futbolista.loc[[2, 3],['nombre', 'posicion']]
```

```python
#Seleccionar las filas en las posiciones, y las columnas por posiciones.
futbolista.iloc[[0, 3], [1]]
```

```python
#realizar cAlculos del promedio de las horas, agrupado por la columna posicion.
futbolista.groupby('posicion')['horas_entrenadas'].mean()
```

```python
#Crear mi primera tabla dinamica
futbolista.pivot_table(index = 'posicion', values = ['horas_entrenadas'], columns = 'liga', aggfunc='max')
```
